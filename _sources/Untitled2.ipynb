```python
#módulos para resolução dos exercícios:
import numpy as np
from scipy.optimize import fsolve
```

```python
#Definição das variáveis

Kh = 3.4e-2 # mol/L atm
Ka1 = 4.5e-7 
Ka2 = 7.0e-11


# pressão de CO2 é obtida a partir da razão de mistura informada. Considerado que o sistema está a 1 atm, 
# 360 ppm de CO2 equivalem a 360E-6 atm de CO2 na atmosfera.
pCO2 = 360e-6 # atm (pressão parcial de CO2 no vapor)


Ka = 20
Kh_hno3 = 2.1e5
pHNO3 = 10e-9


def equations(vars):
    CO2, HCO3, H, HNO3, NO3 = vars
    eq1 = Kh*pCO2 -CO2 
    eq2 = Ka1*CO2 - g**2*HCO3*H
    eq3 = Ka*HNO3 - g**2*NO3*H
    eq4 = Kh_hno3*pHNO3 - HNO3
    eq5 = HCO3 + NO3 - H
    return [eq1, eq2, eq3, eq4, eq5]
```

```python
erro = 10
g = 1

while erro > 1e-10:
    CO2, HCO3, H1, HNO3,NO3 =  fsolve(equations, (1e-2, 1e-6,1e-3, 1e-20,1e-3 ))
    I = 0.5 * (H1 + HCO3 + NO3)
    g = 10**(-0.509*(I**0.5/(1+I**0.5) - 0.3*I))
    CO2, HCO3, H2, HNO3,NO3 =  fsolve(equations, (1e-2, 1e-6,1e-3, 1e-20,1e-3))
    erro = H2 - H1
    print(erro)

print('Solução com fsolve= ' +  format(CO2 , ' 6.5e') + format(H2 , ' 6.5e'))
print('concentração de nitrato=', NO3, 'mol/l' )

pH = -np.log10(H2)
print('pH = ', pH)    
```
