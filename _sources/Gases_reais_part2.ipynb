## Exemplo: Cálculo do volume de uma gás que obedece a equação de van der Waals 

<font size = "4">
Estime o volume molar do $ CO_2 $, a 500 K e 100 atm, admitindo que o gás se comporta como um gás de van der Waals. <font>
    
### Solução
Como vimos na seção anterior, as raízes da equação abaixo devem ser calculadas para obtermos o volume do $CO_2$.
    
$ pV_m^3 - (pb+RT)V_m^2 + aV_m -ab = 0 $

    
Inicialmente ela será calculada por um método iterativo. Posteriormente usando a rotina fsolve(). 
O método iterativo demanda um estimador inicial para o valor do volume, e para evitar erros será usado o método de Newton-Raphson. De acordo com o valor inicial estimado, o processo iterativo pode convergir mais ou menos rápido. Neste caso é uma boa ideia usar o volume molar do gás ideal como estimador inicial do volume. O processo todo pode ser representado da seguinte forma:
    
    
1. **Estimador**: Inicie escolhendo um estimador $( x_0 )$.

2. **Fórmula iterativa**: use a seguinte equação para iterações:
$[ x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)} ]$
Onde $ ( f(x) )$ é um valor da função polinomial e $( f'(x) )$ é sua primeria derivada.

3. **Critério de convergência**: Continue as iterações até $( |x_{n+1} - x_n| ) $ ser próximo de zero.

4. **Solução**: O valor final de $( x ) $ obtido após convergência é uma solução aproximada do problema.    
 <font>

```python
#Definindo as variáveis segundo dados da literatura.

a = 3.610       # dm^6 atm mol^-1
b = 4.29e-2     # dm^3 mol^-1
R = 0.082057    #L · atm · K−1 · mol−1  
T = 500         # K
p = 100         # atm



# 1) Criar um vetor com os coeficientes do polinômio
# os valores dos coeficientes foram calculados dentro do vetor 
# coeficientes ordenados a partir do coeficiente de maior ordem para o de menor ordem

coef = np.array([p, -(p*b+R*T), a, -a*b])


def vdW(V):
    
    return coef[0]*V**3 + coef[1]*V**2 + coef[2]*V**1 + coef[3]
```

```python


Vi = R*T/p
erro = vdW(Vi) - Vi

x = Vi # inicializando o contador

while erro > 0.01: #se este valor for muito pequeno a conta não converge
    
    diff = 3*coef[0]*x**2 + 2*coef[1]*x + coef[2] #escrevendo a derivada da função
    
    vdW_n = vdW(x) # Guardando o valor de V na atual iteração
    
    x = x - vdW(x)/ diff
    
    erro = abs(vdW(x) - vdW_n)
    
print("Volume do gás= " + format(x , ' 5.3e'), "L")
```

<font size ="4"> Usaremos agora a rotina fsolve(). Esta rotina é a implementação de um método também iterativo, o método de Powell. Além do cálculo de raízes, pode ser usado para encontrar a solução de sistemas de equações não-lineares. Este método também faz uso de um estimador inicial, que neste caso será o volume do gás ideal.
    
 <font>

```python
# Reiniciando o estimador

Vi = R*T/p
Vol = fsolve(vdW,Vi)

print("Volume do gás= ", Vol, "L")
```
